````python
import time
from openai import OpenAI

class OpenAIResponsesWrapper:
    def __init__(self, model_id="gpt-5-mini"):
        """Wrapper for OpenAI's Responses API with code interpreter support."""
        self.client = OpenAI()  # Assumes OPENAI_API_KEY is set
        self.model_id = model_id
    
    def generate_with_code_execution(self, prompt: str, poll_interval=1):
        """Send a prompt with code interpreter enabled and return text output."""
        response = self.client.responses.create(
            model=self.model_id,
            input=prompt,
            tools=[{
                "type": "code_interpreter",
                "container": {"type": "auto"}
            }],
        )

        # Poll until finished
        while response.status not in ("completed", "failed"):
            time.sleep(poll_interval)
            response = self.client.responses.retrieve(response.id)

        if response.status != "completed":
            raise RuntimeError(f"Run failed with status: {response.status}")

        # Extract text outputs
        outputs = []
        for item in response.output or []:
            if item.type == "message":
                for c in item.content:
                    if c.type == "output_text":
                        outputs.append(c.text)

        return "\n".join(outputs) if outputs else "No text output found."
        

# Example usage
if __name__ == "__main__":
    wrapper = OpenAIResponsesWrapper("gpt-5-mini")
    result = wrapper.generate_with_code_execution(prompt)
    print(result)
````

````python
import base64
import time
import json
from io import BytesIO
from PIL import Image
from openai import OpenAI

class OpenAIWrapper_2:
    def __init__(self, model_id="gpt-4o-mini", code_exec=False, poll_interval=1):
        """
        Wrapper for OpenAI multimodal models with optional code execution.
        
        Args:
            model_id (str): Model to use (e.g. gpt-4o-mini, gpt-5-mini).
            code_exec (bool): Whether to enable code interpreter tool.
            poll_interval (int): Poll interval (s) for code exec.
        """
        self.client = OpenAI()
        self.model_id = model_id
        self.code_exec = code_exec
        self.poll_interval = poll_interval

    def _image_to_base64(self, image: Image.Image) -> str:
        """Convert PIL Image to base64 string for OpenAI API."""
        buffered = BytesIO()
        image.save(buffered, format="PNG")
        return base64.b64encode(buffered.getvalue()).decode("utf-8")

    def generate(self, image: Image.Image, instruction: str) -> str:
        """
        Generate response given an image and instruction.
        Uses code execution if self.code_exec=True.
        """
        content = [{"type": "text", "text": instruction}]

        if image is not None:
            image_b64 = self._image_to_base64(image)
            content.append({
                "type": "image_url",
                "image_url": {"url": f"data:image/png;base64,{image_b64}"}
            })

        if self.code_exec:
            # Responses API input must use input_text / input_image
            content = [{"type": "input_text", "text": instruction}]
            if image is not None:
                image_b64 = self._image_to_base64(image)
                content.append({"type": "input_image", "image_data": image_b64})

            response = self.client.responses.create(
                model=self.model_id,
                input=[{"role": "user", "content": content}],
                tools=[{"type": "code_interpreter", "container": {"type": "auto"}}],
            )


            # Poll until finished
            while response.status not in ("completed", "failed"):
                time.sleep(self.poll_interval)
                response = self.client.responses.retrieve(response.id)

            if response.status != "completed":
                raise RuntimeError(f"Run failed with status: {response.status}")

            # Collect text outputs
            outputs = []
            for item in response.output or []:
                if item.type == "message":
                    for c in item.content:
                        if c.type == "output_text":
                            outputs.append(c.text)

            return "\n".join(outputs) if outputs else "No text output found."

        else:
            # Chat Completions (text + image only)
            response = self.client.chat.completions.create(
                model=self.model_id,
                messages=[{"role": "user", "content": content}],
            )
            return response.choices[0].message.content.strip()
````

````python
import os
import json
from PIL import Image
import numpy as np
from openai import OpenAI
import time

# ----------------------------
# Local Python functions
# ----------------------------
def compute_average_color(image_path: str) -> str:
    """Compute average RGB color of an image file."""
    image = Image.open(image_path)
    image_np = np.array(image)
    avg_color = np.mean(image_np, axis=(0, 1))
    return f"rgb({int(avg_color[0])}, {int(avg_color[1])}, {int(avg_color[2])})"

def resize_image(image_path: str, width: int, height: int) -> str:
    """Resize image and save a new file, creating directories if needed."""
    image = Image.open(image_path)
    image = image.resize((width, height))
    
    # Ensure directory exists
    base_dir = os.path.dirname(image_path)
    if base_dir and not os.path.exists(base_dir):
        os.makedirs(base_dir)
    
    new_path = os.path.join(base_dir, f"resized_{width}x{height}_{os.path.basename(image_path)}")
    image.save(new_path)
    return new_path

# Map of tool names to local Python functions
LOCAL_FUNCTIONS = {
    "compute_average_color": compute_average_color,
    "resize_image": resize_image
}

# ----------------------------
# OpenAI client setup
# ----------------------------
client = OpenAI()

# ----------------------------
# Define tools for the model
# ----------------------------
tools = [
    {
        "type": "function",
        "function": {
            "name": "compute_average_color",
            "description": "Compute the average RGB color of an image file.",
            "parameters": {
                "type": "object",
                "properties": {
                    "image_path": {"type": "string", "description": "Path to the image file"}
                },
                "required": ["image_path"]
            }
        }
    },
    {
        "type": "function",
        "function": {
            "name": "resize_image",
            "description": "Resize an image to the specified width and height.",
            "parameters": {
                "type": "object",
                "properties": {
                    "image_path": {"type": "string"},
                    "width": {"type": "integer"},
                    "height": {"type": "integer"}
                },
                "required": ["image_path", "width", "height"]
            }
        }
    }
]

# ----------------------------
# User prompt / instruction
# ----------------------------
instruction = """
Analyze the image 'images/stego_image_indexed_5.png', return its average color using compute_average_color,
then resize it to 100x100 using resize_image.
"""

# ----------------------------
# Initial API call with tools
# ----------------------------
response = client.chat.completions.create(
    model="gpt-4o-mini",  # Use an appropriate model (e.g., gpt-4o-mini or gpt-3.5-turbo)
    messages=[
        {"role": "user", "content": instruction}
    ],
    tools=tools,
    tool_choice="auto"
)

# Get the initial response
completion = response.choices[0].message

# Check if the model requested tool calls
tool_calls = completion.tool_calls
if tool_calls:
    # Prepare messages to include the initial user message and tool calls
    messages = [
        {"role": "user", "content": instruction},
        completion  # Include the assistant's response with tool calls
    ]

    # Execute local tools and prepare tool outputs
    for tool_call in tool_calls:
        func_name = tool_call.function.name
        args = json.loads(tool_call.function.arguments)
        if func_name in LOCAL_FUNCTIONS:
            try:
                # Execute local function
                result = LOCAL_FUNCTIONS[func_name](**args)
                # Append tool result to messages
                messages.append({
                    "role": "tool",
                    "content": str(result),
                    "tool_call_id": tool_call.id
                })
            except Exception as e:
                # Handle errors in tool execution
                messages.append({
                    "role": "tool",
                    "content": f"Error executing {func_name}: {str(e)}",
                    "tool_call_id": tool_call.id
                })

    # Submit tool outputs back to the model
    final_response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages,
        tools=tools,
        tool_choice="auto"
    )

    # Print the final output
    # for message in final_response.choices[0].message.content:
    #     print(message)
    print(final_response.choices[0].message.content)

else:
    # If no tool calls, print the direct response
    print(completion.content)
````

````python
import os
import json
from PIL import Image
import numpy as np
from openai import OpenAI
import base64
from io import BytesIO
from typing import List, Dict, Any, Optional, Union, Callable

class OpenAIWrapper:
    def __init__(
        self, 
        model_id: str = "gpt-4o-mini", 
        api_key: Optional[str] = None,
        tools: Optional[List[Dict]] = None,
        local_functions: Optional[Dict[str, Callable]] = None
    ):
        """
        Wrapper for OpenAI's GPT models with tool calling and image processing capabilities.
        
        Args:
            model_id (str): Which OpenAI model to use, e.g., "gpt-4o-mini", "gpt-4o"
            api_key (str, optional): OpenAI API key. If None, uses environment variable OPENAI_API_KEY
            tools (list, optional): List of tool definitions for OpenAI function calling
            local_functions (dict, optional): Dictionary mapping function names to local Python functions
        """
        self.model_id = model_id
        self.client = OpenAI(api_key=api_key) if api_key else OpenAI()
        
        # Initialize tools and local functions
        self.tools = tools if tools is not None else []
        self.local_functions = local_functions if local_functions is not None else {}
        
        # Add default tools if none provided
        if not self.tools:
            self._setup_default_tools()

    def _setup_default_tools(self):
        """Set up default tools if none are provided."""
        self.tools = [
            {
                "type": "function",
                "function": {
                    "name": "compute_average_color",
                    "description": "Compute the average RGB color of an image file.",
                    "parameters": {
                        "type": "object",
                        "properties": {
                            "image_path": {"type": "string", "description": "Path to the image file"}
                        },
                        "required": ["image_path"]
                    }
                }
            },
            {
                "type": "function",
                "function": {
                    "name": "resize_image",
                    "description": "Resize an image to the specified width and height.",
                    "parameters": {
                        "type": "object",
                        "properties": {
                            "image_path": {"type": "string"},
                            "width": {"type": "integer"},
                            "height": {"type": "integer"}
                        },
                        "required": ["image_path", "width", "height"]
                    }
                }
            }
        ]
        
        # Add default local functions
        self.local_functions = {
            "compute_average_color": self._compute_average_color,
            "resize_image": self._resize_image
        }

    def _image_to_base64(self, image: Image.Image) -> str:
        """Convert PIL Image to base64 string for OpenAI API."""
        buffered = BytesIO()
        image.save(buffered, format="PNG")
        return base64.b64encode(buffered.getvalue()).decode("utf-8")

    def _compute_average_color(self, image_path: str) -> str:
        """Compute average RGB color of an image file."""
        image = Image.open(image_path)
        image_np = np.array(image)
        avg_color = np.mean(image_np, axis=(0, 1))
        return f"rgb({int(avg_color[0])}, {int(avg_color[1])}, {int(avg_color[2])})"

    def _resize_image(self, image_path: str, width: int, height: int) -> str:
        """Resize image and save a new file, creating directories if needed."""
        image = Image.open(image_path)
        image = image.resize((width, height))
        
        # Ensure directory exists
        base_dir = os.path.dirname(image_path)
        if base_dir and not os.path.exists(base_dir):
            os.makedirs(base_dir)
        
        new_path = os.path.join(base_dir, f"resized_{width}x{height}_{os.path.basename(image_path)}")
        image.save(new_path)
        return new_path

    def add_tool(self, tool_name: str, tool_function: Callable, tool_definition: Dict[str, Any]):
        """
        Add a custom tool to the wrapper.
        
        Args:
            tool_name (str): Name of the tool function
            tool_function (callable): The Python function to execute
            tool_definition (dict): OpenAI tool definition
        """
        self.local_functions[tool_name] = tool_function
        self.tools.append(tool_definition)

    def generate_with_tools(
        self, 
        instruction: str, 
        image: Optional[Image.Image] = None,
        custom_tools: Optional[List[Dict]] = None,
        custom_functions: Optional[Dict[str, Callable]] = None,
        max_iterations: int = 3
    ) -> str:
        """
        Generate response using tool calling functionality.
        
        Args:
            instruction (str): The user's instruction/prompt
            image (PIL.Image, optional): Image to process
            custom_tools (list, optional): Custom tools to use for this call only
            custom_functions (dict, optional): Custom functions to use for this call only
            max_iterations (int): Maximum number of tool call iterations
            
        Returns:
            str: The final response from the model
        """
        # Use custom tools/functions if provided, otherwise use instance ones
        use_tools = custom_tools if custom_tools is not None else self.tools
        use_functions = custom_functions if custom_functions is not None else self.local_functions
        
        # Prepare messages
        messages = [{"role": "user", "content": []}]
        
        # Add text instruction
        messages[0]["content"].append({"type": "text", "text": instruction})
        
        # Add image if provided
        if image is not None:
            if not isinstance(image, Image.Image):
                image = image.convert("RGB")
            image_b64 = self._image_to_base64(image)
            messages[0]["content"].append({
                "type": "image_url",
                "image_url": {"url": f"data:image/png;base64,{image_b64}"}
            })
        
        # Iterate for tool calls
        for iteration in range(max_iterations):
            # Call the API
            response = self.client.chat.completions.create(
                model=self.model_id,
                messages=messages,
                tools=use_tools if iteration == 0 else None,  # Only send tools on first call
                tool_choice="auto" if iteration == 0 else None
            )
            
            completion = response.choices[0].message
            
            # Check if tool calls are requested
            tool_calls = completion.tool_calls
            if not tool_calls:
                return completion.content
            
            # Add assistant's response to messages
            messages.append(completion)
            
            # Execute tool calls
            for tool_call in tool_calls:
                func_name = tool_call.function.name
                args = json.loads(tool_call.function.arguments)
                
                if func_name in use_functions:
                    try:
                        # Execute local function
                        result = use_functions[func_name](**args)
                        # Append tool result to messages
                        messages.append({
                            "role": "tool",
                            "content": str(result),
                            "tool_call_id": tool_call.id
                        })
                    except Exception as e:
                        # Handle errors in tool execution
                        messages.append({
                            "role": "tool",
                            "content": f"Error executing {func_name}: {str(e)}",
                            "tool_call_id": tool_call.id
                        })
                else:
                    messages.append({
                        "role": "tool",
                        "content": f"Function {func_name} not available",
                        "tool_call_id": tool_call.id
                    })
        
        # If we reach max iterations without final response, get one last response
        final_response = self.client.chat.completions.create(
            model=self.model_id,
            messages=messages
        )
        
        return final_response.choices[0].message.content

    def generate_simple(
        self, 
        image: Image.Image, 
        instruction: str
    ) -> str:
        """
        Simple generation without tool calling.
        
        Args:
            image (PIL.Image): Image to process
            instruction (str): The user's instruction/prompt
            
        Returns:
            str: The response from the model
        """
        if not isinstance(image, Image.Image):
            image = image.convert("RGB")

        image_b64 = self._image_to_base64(image)

        response = self.client.chat.completions.create(
            model=self.model_id,
            messages=[
                {
                    "role": "user",
                    "content": [
                        {"type": "text", "text": instruction},
                        {
                            "type": "image_url",
                            "image_url": {"url": f"data:image/png;base64,{image_b64}"}
                        }
                    ],
                }
            ],
        )

        return response.choices[0].message.content.strip()


# Example usage
if __name__ == "__main__":
    # Define custom tools and functions
    custom_tools = [
        {
            "type": "function",
            "function": {
                "name": "analyze_image_histogram",
                "description": "Analyze the color histogram of an image.",
                "parameters": {
                    "type": "object",
                    "properties": {
                        "image_path": {"type": "string", "description": "Path to the image file"}
                    },
                    "required": ["image_path"]
                }
            }
        }
    ]
    
    def custom_histogram_analysis(image_path: str) -> str:
        """Custom function to analyze image histogram."""
        image = Image.open(image_path)
        histogram = image.histogram()
        return f"Histogram analysis complete. Total pixels: {sum(histogram)}"
    
    custom_functions = {
        "analyze_image_histogram": custom_histogram_analysis
    }
    
    # Initialize wrapper with custom tools
    wrapper = OpenAIWrapper(
        model_id="gpt-4o-mini",
        tools=custom_tools,
        local_functions=custom_functions
    )
    
    print("Available tools:", [tool['function']['name'] for tool in wrapper.tools])
    print("Available functions:", list(wrapper.local_functions.keys()))
    
    # Example usage with custom tools
    instruction = "Analyze the histogram of image 'test_image.png'"
    
    try:
        result = wrapper.generate_with_tools(instruction)
        print("Tool calling result:", result)
    except Exception as e:
        print(f"Example failed (expected without actual image): {e}")
    
    # Example with per-call custom tools
    per_call_tools = [
        {
            "type": "function",
            "function": {
                "name": "get_image_metadata",
                "description": "Get metadata of an image file.",
                "parameters": {
                    "type": "object",
                    "properties": {
                        "image_path": {"type": "string", "description": "Path to the image file"}
                    },
                    "required": ["image_path"]
                }
            }
        }
    ]
    
    def get_metadata(image_path: str) -> str:
        """Get image metadata."""
        image = Image.open(image_path)
        return f"Size: {image.size}, Format: {image.format}, Mode: {image.mode}"
    
    per_call_functions = {
        "get_image_metadata": get_metadata
    }
    
    # Use per-call custom tools
    try:
        result = wrapper.generate_with_tools(
            instruction="Get metadata for image 'test_image.png'",
            custom_tools=per_call_tools,
            custom_functions=per_call_functions
        )
        print("Per-call tool result:", result)
    except Exception as e:
        print(f"Per-call example failed (expected): {e}")
````
